### (a) $\mathcal{W}e(\alpha,1)$ 

Suponiendo parámetro de forma $\alpha$ y escala $\lambda=1$. Sabemos que la función de distribución acumulada es: $F(x) = 1 - e^{-x^\alpha}$ para $x \ge 0$.

Paso 1: Generar $U \sim U(0,1)$

Paso 2: $F(x) = U$ y despejando $x$

$1 - e^{-x^\alpha} = U $ 
$\implies$ $e^{-x^\alpha} = 1 - U$ 
$\implies$ $ -x^{\alpha} = \ln(1 - U)$ 
$\implies$ $x = \left( -\ln(1 - U) \right)^{1/\alpha}$

Función de densidad: $f(x) = \alpha x^{\alpha-1} e^{-x^\alpha}$

```python
import numpy as np
import random
import matplotlib.pyplot as plt
```

```python
def weibull(alpha,n):
    muestra=[]
    for i in range(n):
        u=random.random()
        x=(-np.log(1-u))**(1/alpha)
        muestra.append(float(x))
    return muestra
n=10000
alpha=2
muestra_w=weibull(alpha,n)
print(f"Media:{np.mean(muestra_w)}")
print(f"Varianza:{np.var(muestra_w)}")
print(f"Desviacion estandar:{np.std(muestra_w)}")
xt=np.linspace(0,max(muestra_w),1000)
fxt=alpha*xt**(alpha-1)*np.exp(-xt**alpha)
Fxt=1-np.exp(-xt**alpha)

plt.hist(muestra_w,bins=100,density=True,alpha=0.5,label='Histograma')
plt.plot(xt,fxt,label='f(x)')
plt.legend()
plt.show()
plt.plot(np.sort(muestra_w),np.arange(1,n+1)/n,'.',alpha=0.1,label='Muestra ordenada')
plt.plot(xt,Fxt,label='F(x)')
plt.legend()
plt.show()
```

### (b) $\mathcal{G}u(0,1)$

Sabemos que la función de distribución acumulada es:$F(x) = e^{-e^{-x}}$ para $x \in \mathbb{R}$

Paso 1:Sea $U \sim U(0,1)$

Paso 2: Escribir $F(x) = U$ y despejar $x$

$e^{-e^{-x}} = U$ $\implies$
$-e^{-x} = \ln(U)$ $\implies$
$e^{-x} = -\ln(U)$ $\implies$
$-x = \ln(-\ln(U))$ $\implies$
$x = -\ln(-\ln(U))$ 

Función de densidad: $f(x) = e^{-x} e^{-e^{-x}}$

```python
def gumbel(n):
    muestra=[]
    for i in range(n):
        u=random.random()
        if u==0.0: u=1e-10 
        x=-np.log(-np.log(u))
        muestra.append(float(x))
    return muestra
n=100000
muestra_g=gumbel(n)
print(f"Media:{np.mean(muestra_g)}")
print(f"Varianza:{np.var(muestra_g)}")
print(f"Desviacion estandar:{np.std(muestra_g)}")
xt=np.linspace(min(muestra_g),max(muestra_g),1000)
fxt=np.exp(-xt)*np.exp(-np.exp(-xt))
Fxt=np.exp(-np.exp(-xt))

plt.hist(muestra_g,bins=100,density=True,alpha=0.5,label='Histograma')
plt.plot(xt,fxt,label='f(x)')
plt.legend()
plt.show()
plt.plot(np.sort(muestra_g),np.arange(1,n+1)/n,'.',alpha=0.1,label='Muestra ordenada')
plt.plot(xt,Fxt,label='F(x)')
plt.legend()
plt.show()
```

### (c) $\mathcal{C}(0,1)$
Sabemos que la función de distribución acumulada es: $F(x) = \frac{1}{\pi} \arctan(x) + \frac{1}{2}$ para $x \in \mathbb{R}$

Paso 1:Sea $U \sim U(0,1)$

Paso 2:Escribir $F(x) = U$ y despejar $x$

$\frac{1}{\pi} \arctan(x) + \frac{1}{2}=U$ $\implies$
$\frac{1}{\pi} \arctan(x)=U-\frac{1}{2}$ $\implies$
$\arctan(x)=\pi\left(U-\frac{1}{2}\right)$ $\implies$
$x=\tan\left(\pi\left(U-\frac{1}{2}\right)\right)$               
Función de densidad: $f(x) = \frac{1}{\pi(1+x^2)}$

```python
import numpy as np
import random
import matplotlib.pyplot as plt
```

```python
def cauchy(n):
    muestra=[]
    for i in range(n):
        u=random.random()
        x=np.tan(np.pi*(u-0.5))
        muestra.append(float(x))
    return muestra
n=10000
muestra_c=cauchy(n)
print(f"Media:{np.mean(muestra_c)}")
print(f"Varianza:{np.var(muestra_c)}")
print(f"Desviacion estandar:{np.std(muestra_c)}")
xt=np.linspace(-15,15,1000)
fxt=1/(np.pi*(1+xt**2))
Fxt=(1/np.pi)*np.arctan(xt)+0.5

plt.hist(muestra_c,bins=1000,density=True,alpha=0.5,label='Histograma')
plt.plot(xt,fxt,label='f(x)')
plt.xlim(-15,15)
plt.legend()
plt.show()
plt.plot(np.sort(muestra_c),np.arange(1,n+1)/n,'.',alpha=0.1,label='Muestra ordenada')
plt.plot(xt,Fxt,label='F(x)')
plt.xlim(-15,15)
plt.legend()
plt.show()
```

### (d) $\mathcal{L}(0,1)$}
Sabemos que la función de distribución acumulada es: 
$F(x) = \frac{1}{2} e^x$ para $x < 0$  y  $F(x) = 1 - \frac{1}{2} e^{-x}$ para $x \ge 0$

Paso 1: Sea $U \sim U(0,1)$

Paso 2: Escribir $F(x)=U$ y despejar $x$

- Para $U<0.5$:

$\frac{1}{2} e^x =U \implies^x = 2 U \implies x =\ln (2U)$
    
- Para $U \ge 0.5$:

$1 - \frac{1}{2} e^{-x} = U \implies \frac{1}{2} e^{-x} = 1 - U \implies x = -\ln(2(1 - U))$

Función de densidad: $f(x) = \frac{1}{2} e^{-|x|}$

```python
import numpy as np
import random
import matplotlib.pyplot as plt
```

```python
def laplace(n):
    muestra=[]
    for i in range(n):
        u=random.random()
        if u<0.5:
            if u==0.0: u=1e-10
            x=np.log(2*u)
        else:
            x=-np.log(2*(1-u))
        muestra.append(float(x))
    return muestra
n=10000
muestra_l=laplace(n)
print(f"Media:{np.mean(muestra_l)}")
print(f"Varianza:{np.var(muestra_l)}")
print(f"Desviacion estandar:{np.std(muestra_l)}")

xt=np.linspace(min(muestra_l),max(muestra_l),1000)
fxt=0.5*np.exp(-np.abs(xt))
Fxt=np.where(xt<0,0.5*np.exp(xt),1-0.5*np.exp(-xt))

plt.hist(muestra_l,bins=100,density=True,alpha=0.5,label='Histograma')
plt.plot(xt,fxt,label='f(x)')
plt.legend()
plt.show()
plt.plot(np.sort(muestra_l),np.arange(1,n+1)/n,'.',alpha=0.1,label='Muestra ordenada')
plt.plot(xt,Fxt,label='F(x)')
plt.legend()
plt.show()
```

### (e) $\mathcal{P}ar(\alpha)$

Suponiendo al parámetro de escala $x_m=1$

Sabemos que la función de distribución acumulada es: 
$F(x) = 1 - x^{-\alpha}$ para $x \ge 1$

Paso 1: Sea $U \sim U(0,1)$

Paso 2: Escribir $F(x) = U$ y despejar $x$

$1 - x^{-\alpha} = U$ $\implies$
$x^{-\alpha} = 1 - U$ $\implies$
$x^\alpha = \frac{1}{1 - U}$ $\implies$
$x = (1 - U)^{-1/\alpha}$

Función de densidad:$f(x) = \alpha x^{-(\alpha+1)}$

```python
import numpy as np
import random
import matplotlib.pyplot as plt
```

```python
def pareto(alpha,n):
    muestra=[]
    for i in range(n):
        u=random.random()
        x=(1-u)**(-1/alpha)
        muestra.append(float(x))
    return muestra
n=10000
alpha=2
muestra_p=pareto(alpha,n)
print(f"Media:{np.mean(muestra_p)}")
print(f"Varianza:{np.var(muestra_p)}")
print(f"Desviacion estandar:{np.std(muestra_p)}")
xt=np.linspace(1,10,1000)
fxt=alpha*xt**(-alpha-1)
Fxt=1-xt**(-alpha)

plt.hist(muestra_p,bins=1000,density=True,alpha=0.5,label='Histograma')
plt.plot(xt,fxt,label='f(x)')
plt.xlim(1,10)
plt.legend()
plt.show()
plt.plot(np.sort(muestra_p),np.arange(1,n+1)/n,'.',alpha=0.1,label='Muestra ordenada')
plt.plot(xt,Fxt,label='F(x)')
plt.xlim(1,10)
plt.legend()
plt.show()
```
